```python
# @title Forced Aligner
# @markdown Enter an Hugging Face Wav2Vec2-based model. Empty string to use the default torchaudio MMS-FA model.
forced_aligner = "NextFire/mms-300m-ForcedAligner-karaoke-ja-Latn" # @param {type:"string"}
```

```python
# @title Source Separator { display-mode: "form", run: "auto" }
# @markdown Run this cell and select a **Source Separator**:
# @markdown - **None**: Disables source separation (unecessary if using `NextFire/mms-300m-ForcedAligner-karaoke-ja-Latn`).
# @markdown - **VocalRemoverSeparator**: Based on the [`vocal-remover`](https://github.com/tsurumeso/vocal-remover) library. Choose this if you're unsure.
# @markdown - **HybridDemucsSeparator**: Uses `torchaudio`'s [Hybrid Demucs model](https://pytorch.org/audio/2.1.0/tutorials/hybrid_demucs_tutorial.html), which is faster but less aggressive.

from yohane.audio import VocalRemoverSeparator, HybridDemucsSeparator

separator_class = None # @param ["VocalRemoverSeparator", "HybridDemucsSeparator", "None"] {type:"raw"}
separator = separator_class() if separator_class is not None else None
```

## Run

When ready, execute the next cells to run the pipeline.


```python
# @title Generate

import logging
from pathlib import Path
from yohane import Yohane
from huggingface_hub import login # Add this import

logging.basicConfig(level="INFO", force=True)
logging.getLogger("httpx").setLevel(logging.WARNING)

# It is recommended to log in to Hugging Face Hub for higher rate limits and faster downloads.
# You can generate an API token here: https://huggingface.co/settings/tokens
# Paste your token here or set it as an environment variable (e.g., in a Colab secret).
# For example: login(token="hf_YOUR_TOKEN_HERE")
# If you are using Colab secrets, you can get the token like this:
# from google.colab import userdata
# login(token=userdata.get('HF_TOKEN')) # Assumes you saved your token as 'HF_TOKEN' secret.
# Uncomment and replace "hf_YOUR_TOKEN_HERE" with your actual token or use Colab secrets:
# # login(token="hf_YOUR_TOKEN_HERE")

song_path = Path(song_filename)

yohane = Yohane(separator=separator, forced_aligner=forced_aligner)
yohane.load_song(song_path)
yohane.load_lyrics(lyrics_area.value)
yohane.extract_vocals()
yohane.force_align()
subs = yohane.make_subs()
```

```python
# @title Save and download karaoke file

from google.colab import files

subs_path = song_path.with_suffix('.ass')
subs.save(subs_path)
files.download(subs_path)
```

```python
# @title Download song file

files.download(song_path)
```
